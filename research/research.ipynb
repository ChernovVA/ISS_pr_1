```python
import os

from sklearn.model_selection import train_test_split

import pandas as pd
import numpy as np

from sklearn.preprocessing import StandardScaler, OrdinalEncoder

from sklearn.pipeline import Pipeline

from sklearn.compose import ColumnTransformer

from sklearn.ensemble import RandomForestRegressor

from sklearn.linear_model import LinearRegression

from catboost import CatBoostRegressor

from sklearn.metrics import (
    mean_absolute_percentage_error,
    mean_absolute_error,
    mean_squared_error
)
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
import mlflow

TRACKING_SERVER_HOST = "127.0.0.1"
TRACKING_SERVER_PORT = 5000

tracking_uri = f"http://{TRACKING_SERVER_HOST}:{TRACKING_SERVER_PORT}"

mlflow.set_tracking_uri(tracking_uri)

from mlflow.models import infer_signature
```

## Загрузка датасета

```python
df = pd.read_pickle('/home/mainuser/ISS/Pr_1/data/clean_dataset.pkl')
df = df.rename(columns={'Selling_Price': 'target'})
df = df.drop(columns=['Car_Name'])
df.info()
```

```python
X = df.drop(columns=['target'])
y = df['target']

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42
)
#Разбивка признаков на категориальные и числовые
cat_features = df.select_dtypes(
    include=['category']
).columns.to_list()

print("Category: ", cat_features)

num_features = df.drop(columns=['target']).select_dtypes(
    include=['number']
).columns.to_list()

print("Numeral: ", num_features)
```

## Создание  baseline-модели


```python
transformer = ColumnTransformer([
    ('num', StandardScaler(), num_features),
    ('cat', OrdinalEncoder(), cat_features)
])

regressor = RandomForestRegressor(random_state=42)
```

```python
pipeline = Pipeline(steps=[('transformer', transformer), 
                           ('model', regressor)])

pipeline.fit(X_train, y_train)
```

```python
y_pred = pipeline.predict(X_test)

mae = mean_absolute_error(y_test, y_pred)
mape = mean_absolute_percentage_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)

print('1. MAE:', mae)
print('2. MSE:', mse)
print('3. MAPE:', mape)
```

```python
with mlflow.start_run(experiment_id=1, run_name="RandomForest_baseline") as run:

    pipeline.fit(X_train, y_train)

    y_pred = pipeline.predict(X_test)

    mae = mean_absolute_error(y_test, y_pred)
    mape = mean_absolute_percentage_error(y_test, y_pred)
    mse = mean_squared_error(y_test, y_pred)

    mlflow.log_metric("MAE", mae)
    mlflow.log_metric("MSE", mse)
    mlflow.log_metric("MAPE", mape)

    input_example = X_train.head(5)

    signature = infer_signature(
        X_train.head(5),
        pipeline.predict(X_train.head(5))
    )

    mlflow.sklearn.log_model(
        pipeline,
        artifact_path="model",
        input_example=input_example,
        signature=signature,
        registered_model_name="car_price_prediction_model"
    )

    mlflow.log_artifact("../requirements.txt")
```

## Добавление полиномиальных признаков

```python
from sklearn.preprocessing import PolynomialFeatures

poly_transformer = ColumnTransformer([
    (
        'num_poly',
        Pipeline([
            ('poly', PolynomialFeatures(degree=2, include_bias=False)), #макс. вторая степень + без столбца единиц
            ('scaler', StandardScaler()) #просто масштабирование
        ]),
        num_features
    ),
    (
        'cat',
        OrdinalEncoder(),
        cat_features
    )
])
```

```python
X_train_fe_sklearn = X_train.copy()
X_train_fe_sklearn = poly_transformer.fit_transform(X_train_fe_sklearn)
```

```python
feature_names = poly_transformer.get_feature_names_out()
print (feature_names)

with open("feature_names.txt", "w", encoding="utf-8") as f:
    for name in feature_names:
        f.write(name + "\n")
```

```python
fe_pipeline = Pipeline([
    ('preprocessor', poly_transformer),
    ('regressor', RandomForestRegressor(random_state=42))
])
```

```python
with mlflow.start_run(run_name="RandomForest_PolynomialFeatures", experiment_id=1) as run:

    fe_pipeline.fit(X_train, y_train)

    y_pred_fe = fe_pipeline.predict(X_test)

    mae_fe = mean_absolute_error(y_test, y_pred_fe)
    mse_fe = mean_squared_error(y_test, y_pred_fe)
    mape_fe = mean_absolute_percentage_error(y_test, y_pred_fe)

    mlflow.log_metric("MAE", mae_fe)
    mlflow.log_metric("MSE", mse_fe)
    mlflow.log_metric("MAPE", mape_fe)

    input_example = X_train.head(5)

    signature = infer_signature(
        X_train.head(5),
        fe_pipeline.predict(X_train.head(5))
    )

    mlflow.sklearn.log_model(
        fe_pipeline,
        artifact_path="model",
        input_example=input_example,
        signature=signature
    )

    mlflow.log_artifact("feature_names.txt")

    print("MAE:", mae_fe)
    print("MSE:", mse_fe)
    print("MAPE:", mape_fe)
```

## Отбор признаков

```python
from mlxtend.feature_selection import SequentialFeatureSelector

# Создаём SequentialFeatureSelector с направлением forward
sfs = SequentialFeatureSelector(
    estimator=RandomForestRegressor(n_estimators=100, random_state=42, n_jobs=1),
    k_features="best",          
    forward=True,               
    floating=False,
    scoring="neg_mean_absolute_error",
    cv=5,
    n_jobs=1,
    verbose=1
)

# Обучаем на расширенном наборе признаков из пункта 10
sfs = sfs.fit(X_train_fe_sklearn, y_train)

# Названия отобранных столбцов и их индексы
selected_indices = list(sfs.k_feature_idx_)
selected_feature_names = [feature_names[i] for i in selected_indices]


print("Отобранные признаки:", selected_feature_names)
print("Индексы отобранных признаков:", selected_indices)
```

```python
with open("selected_feature_names.txt", "w", encoding="utf-8") as f:
    for name in selected_feature_names:
        f.write(name + "\n")

with open("selected_feature_indices.txt", "w", encoding="utf-8") as f:
    for index in selected_indices:
        f.write(str(index) + "\n")
```

```python
fe_pipeline = Pipeline([
    ('preprocessor', poly_transformer),
    ('regressor', RandomForestRegressor(random_state=42))
])
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

feature_selector = ColumnTransformer(
    [
        ("selected", "passthrough", selected_indices)
    ],
    remainder="drop"
) #по факту именно тут мы и отбрасываем все остальные признаки
```

```python
selected_pipeline = Pipeline([
    ('preprocessor', poly_transformer),
    ('feature_selector', feature_selector),
    ('regressor', RandomForestRegressor(random_state=42))
])
```

```python
with mlflow.start_run(run_name="RandomForest_FeatureSelection", experiment_id=1) as run:

    selected_pipeline.fit(X_train, y_train)

    y_pred_selected = selected_pipeline.predict(X_test)

    mae_selected = mean_absolute_error(y_test, y_pred_selected)
    mse_selected = mean_squared_error(y_test, y_pred_selected)
    mape_selected = mean_absolute_percentage_error(y_test, y_pred_selected)

    mlflow.log_metric("MAE", mae_selected)
    mlflow.log_metric("MSE", mse_selected)
    mlflow.log_metric("MAPE", mape_selected)

    input_example = X_train.head(5)

    signature = infer_signature(
        X_train.head(5),
        selected_pipeline.predict(X_train.head(5))
    )

    mlflow.sklearn.log_model(
        selected_pipeline,
        artifact_path="model",
        input_example=input_example,
        signature=signature
    )

    mlflow.log_artifact("selected_feature_names.txt")
    mlflow.log_artifact("selected_feature_indices.txt")

    print("MAE:", mae_selected)
    print("MSE:", mse_selected)
    print("MAPE:", mape_selected)
```

## Подбор гиперпараметров 

```python
import optuna
from sklearn.ensemble import RandomForestRegressor
from sklearn.pipeline import Pipeline
```

```python
def objective(trial):

    n_estimators = trial.suggest_int("n_estimators", 50, 300)
    max_depth = trial.suggest_int("max_depth", 3, 30)
    max_features = trial.suggest_float("max_features", 0.1, 1.0)

    pipeline_tuned = Pipeline([
        ('transform', transformer),
        ('regressor', RandomForestRegressor(
            n_estimators=n_estimators,
            max_depth=max_depth,
            max_features=max_features,
            random_state=42
        ))
    ])

    pipeline_tuned.fit(X_train, y_train)

    y_pred = pipeline_tuned.predict(X_test)

    mape = mean_absolute_percentage_error(y_test, y_pred)

    return mape
```

```python
study = optuna.create_study(direction="minimize")

study.optimize(objective, n_trials=100)
```

```python
print("Лучший MAPE:", study.best_value)
print("Лучшие параметры:")
print(study.best_params)
```

```python
best_params = study.best_params

tuned_pipeline = Pipeline([
    ('transform', transformer),
    ('regressor', RandomForestRegressor(
        **best_params,
        random_state=42
    ))
])
```

```python
with mlflow.start_run(run_name="RandomForest_HyperparameterTuning", experiment_id=1) as run:

    tuned_pipeline.fit(X_train, y_train)

    y_pred_tuned = tuned_pipeline.predict(X_test)

    mae_tuned = mean_absolute_error(y_test, y_pred_tuned)
    mse_tuned = mean_squared_error(y_test, y_pred_tuned)
    mape_tuned = mean_absolute_percentage_error(y_test, y_pred_tuned)


    mlflow.log_params(best_params)

    mlflow.log_metric("MAE", mae_tuned)
    mlflow.log_metric("MSE", mse_tuned)
    mlflow.log_metric("MAPE", mape_tuned)

    input_example = X_train.head(5)

    signature = infer_signature(
        X_train.head(5),
        tuned_pipeline.predict(X_train.head(5))
    )

    mlflow.sklearn.log_model(
        tuned_pipeline,
        artifact_path="model",
        input_example=input_example,
        signature=signature
    )

    print("MAE:", mae_tuned)
    print("MSE:", mse_tuned)
    print("MAPE:", mape_tuned)

    print("Лучшие параметры:", best_params)
```
